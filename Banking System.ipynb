```python
import random
import datetime

all_accounts = {}


def make_acc_num():
    while True:
        num = random.randint(100000, 999999)
        if num not in all_accounts:
            return num


def main():
    while True:
        print("\n=======================")
        print("      BANK SYSTEM      ")
        print("=======================")
        print("1. Create Account")
        print("2. Login")
        print("3. Exit")

        choice = input("Enter choice (1-3): ")

        if choice == "1":
            new_account()
        elif choice == "2":
            login()
        elif choice == "3":
            print("Goodbye!")
            break


def user_menu(acc_num):
    while True:
        print("\n-----------------------")
        print("       USER MENU       ")
        print("-----------------------")
        print("1. Check Balance")
        print("2. Deposit Money")
        print("3. Withdraw Money")
        print("4. Transfer Money")
        print("5. View History")
        print("6. Change PIN")
        print("7. Logout")

        choice = input("Enter choice (1-7): ")

        if choice == "1":
            show_balance(acc_num)
        elif choice == "2":
            add_money(acc_num)
        elif choice == "3":
            get_money(acc_num)
        elif choice == "4":
            send_money(acc_num)
        elif choice == "5":
            show_history(acc_num)
        elif choice == "6":
            change_pin(acc_num)
        elif choice == "7":
            print("Logging out...")
            break


def new_account():
    print("\n--- NEW ACCOUNT ---")
    name = input("Enter your name: ")
    phone = input("Enter phone number: ")
    pin = input("Enter 4-digit PIN: ")

    acc_num = make_acc_num()
    today = str(datetime.datetime.now())

    all_accounts[acc_num] = [name, phone, pin, 0.0, []]
    all_accounts[acc_num][4].append(today + " - Account created")

    print("\nAccount made successfully!")
    print("Your Account Number is:", acc_num)


def login():
    print("\n--- LOGIN ---")
    acc_num = int(input("Enter Account Number: "))

    if acc_num not in all_accounts:
        print("Account not found!")
        return

    pin_input = input("Enter PIN: ")

    if all_accounts[acc_num][2] != pin_input:
        print("Wrong PIN!")
        return

    print("\nWelcome", all_accounts[acc_num][0])
    user_menu(acc_num)


def show_balance(acc_num):
    print("\nYour Balance is: $", all_accounts[acc_num][3])


def add_money(acc_num):
    amt = float(input("\nEnter deposit amount: "))
    if amt <= 0:
        print("Invalid amount!")
        return

    all_accounts[acc_num][3] += amt
    today = str(datetime.datetime.now())
    all_accounts[acc_num][4].append(today + " - Deposited: $" + str(amt) + " | Balance: $" + str(all_accounts[acc_num][3]))
    print("Deposited $", amt)


def get_money(acc_num):
    amt = float(input("\nEnter withdrawal amount: "))

    if amt > all_accounts[acc_num][3]:
        print("Not enough money!")
        return

    all_accounts[acc_num][3] -= amt
    today = str(datetime.datetime.now())
    all_accounts[acc_num][4].append(today + " - Withdrew: $" + str(amt) + " | Balance: $" + str(all_accounts[acc_num][3]))
    print("Withdrew $", amt)


def send_money(acc_num):
    target = int(input("\nEnter receiver account number: "))
    amt = float(input("Enter transfer amount: "))

    if target not in all_accounts or target == acc_num or amt > all_accounts[acc_num][3]:
        print("Transfer failed! Check account number or balance.")
        return

    all_accounts[acc_num][3] -= amt
    all_accounts[target][3] += amt
    today = str(datetime.datetime.now())

    all_accounts[acc_num][4].append(today + " - Sent $" + str(amt) + " to " + str(target))
    all_accounts[target][4].append(today + " - Received $" + str(amt) + " from " + str(acc_num))
    print("Money sent successfully!")


def show_history(acc_num):
    print("\n--- HISTORY ---")
    logs = all_accounts[acc_num][4]
    for i in range(len(logs)):
        print(str(i + 1) + ". " + logs[i])


def change_pin(acc_num):
    print("\n--- CHANGE PIN ---")
    old_pin = input("Enter current PIN: ")

    if old_pin != all_accounts[acc_num][2]:
        print("Wrong PIN!")
        return

    new_pin = input("Enter new 4-digit PIN: ")
    confirm_pin = input("Confirm new PIN: ")

    if new_pin != confirm_pin:
        print("PINs do not match!")
        return

    all_accounts[acc_num][2] = new_pin
    today = str(datetime.datetime.now())
    all_accounts[acc_num][4].append(today + " - PIN changed")
    print("PIN changed successfully!")


main()
```
